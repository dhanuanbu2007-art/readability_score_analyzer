```python
import gradio as gr
import textstat

def analyze_readability(text):

    if not text.strip():
        return "Please enter some text to analyze."

    try:
        words = textstat.lexicon_count(text, removepunct=True)
        sentences = textstat.sentence_count(text)
        syllables = textstat.syllable_count(text)
        characters = textstat.char_count(text, ignore_spaces=True)

        flesch_score = textstat.flesch_reading_ease(text)
        grade_level = textstat.flesch_kincaid_grade(text)

        if flesch_score >= 70:
            difficulty = "Easy to Read"
        elif flesch_score >= 50:
            difficulty = "Moderately Difficult"
        else:
            difficulty = "Difficult to Read"

        result = f"""
## 📊 Readability Analysis Result

**Readability Level:** {difficulty}

**Flesch Reading Ease Score:** {flesch_score:.2f}

**Estimated Grade Level:** {grade_level:.2f}

### Text Statistics

- Total Words: {words}
- Total Sentences: {sentences}
- Total Syllables: {syllables}
- Total Characters: {characters}

### Interpretation

A higher Flesch Reading Ease score generally indicates
that the text is easier to read.

The grade level estimates the educational level
needed to understand the text.
"""
        return result

    except Exception as e:
        return f"Analysis failed: {e}"


demo = gr.Interface(
    fn=analyze_readability,

    inputs=gr.Textbox(
        lines=8,
        label="Enter Your Text",
        placeholder="Paste a paragraph here to analyze its readability..."
    ),

    outputs=gr.Markdown(
        label="Readability Report"
    ),

    title="📖 Readability Score Analyzer",

    description="Analyze text complexity, readability score, and estimated grade level.",

    examples=[
        ["The sun is bright. I like to play outside. It is a beautiful day."],
        ["Natural language processing is a branch of artificial intelligence that enables computers to understand, interpret, and generate human language."],
        ["Education plays an important role in developing knowledge, improving skills, and creating opportunities for individuals."]
    ]
)

demo.launch(share=True)
```
